## 1. ¿Qué es un módulo?

Un **módulo** es simplemente un archivo `.py` con funciones, clases o variables que podés reutilizar en otros archivos.

Python ya viene con una enorme **biblioteca estándar** de módulos listos para usar.

````python
# Importar un módulo completo
import math

print(math.pi)          # Constante pi
print(math.sqrt(16))    # Raíz cuadrada
print(math.factorial(5))  # Factorial
````

````python
# Importar solo lo que necesitás
from math import pi, sqrt

print(pi)
print(sqrt(25))
````

````python
# Importar con alias (nombre alternativo)
import math as m

print(m.log(100, 10))   # log base 10 de 100
````

````python
import yfinance as yf
apple = yf.Ticker("AAPL")
print(apple.info)
````

## 2. ¿Qué es un paquete?

Un **paquete** es una carpeta que contiene múltiples módulos relacionados, más un archivo especial llamado `__init__.py`.

```
mi_paquete/
│
├── __init__.py       ← hace que la carpeta sea un paquete
├── matematica.py
└── texto.py
```

Los paquetes te permiten organizar código grande en partes más manejables.

## 3. Módulos importantes de la biblioteca estándar

### `os` — Sistema operativo

````python
import os

# Directorio actual
print("Directorio actual:", os.getcwd())

# Listar archivos
archivos = os.listdir('.')
print("Archivos:", archivos[:5])  # primeros 5

# Variables de entorno
home = os.environ.get('HOME', 'no disponible')
print("Home:", home)
````

### `random` — Números aleatorios

````python
import random

# Número entero aleatorio entre 1 y 10
print(random.randint(1, 10))

# Número flotante entre 0 y 1
print(random.random())

# Elegir un elemento al azar de una lista
colores = ['rojo', 'verde', 'azul', 'amarillo']
print(random.choice(colores))

# Mezclar una lista
numeros = list(range(1, 11))
random.shuffle(numeros)
print(numeros)
````

### `json` — Leer y escribir JSON

````python
import json

# Diccionario Python → JSON string
datos = {
    "nombre": "Ana",
    "edad": 22,
    "materias": ["Python", "Bases de Datos"]
}

json_string = json.dumps(datos, indent=2, ensure_ascii=False)
print(json_string)

# JSON string → diccionario Python
recuperado = json.loads(json_string)
print(type(recuperado))
print(recuperado['nombre'])
````

### `collections` — Estructuras de datos especiales

````python
from collections import Counter, defaultdict, namedtuple

# Counter: contar elementos
palabras = ['python', 'java', 'python', 'c++', 'python', 'java']
conteo = Counter(palabras)
print("Conteo:", conteo)
print("Más común:", conteo.most_common(2))

# defaultdict: diccionario con valor por defecto
grupo = defaultdict(list)
estudiantes = [('A', 'Juan'), ('B', 'Ana'), ('A', 'Luis'), ('B', 'Maria')]
for comision, nombre in estudiantes:
    grupo[comision].append(nombre)
print(dict(grupo))

# namedtuple: tupla con nombres de campos
Punto = namedtuple('Punto', ['x', 'y'])
p = Punto(3, 4)
print(f"Punto: x={p.x}, y={p.y}")
````

## 5. Crear tu propio paquete

Vamos a simular la creación de un paquete `utilidades` con dos módulos.

**Estructura:**
```
utilidades/
├── __init__.py
├── numeros.py
└── texto.py
```

````python
import os

# Crear la estructura del paquete
os.makedirs('utilidades', exist_ok=True)

# __init__.py — se ejecuta al importar el paquete
with open('utilidades/__init__.py', 'w') as f:
    f.write('''
# Paquete utilidades
# Importar desde los submódulos para facilitar el acceso
from .numeros import es_par, factorial
from .texto import capitalizar_palabras, contar_vocales
''')

# numeros.py
with open('utilidades/numeros.py', 'w') as f:
    f.write('''
def es_par(n):
    """Retorna True si n es par."""
    return n % 2 == 0

def factorial(n):
    """Calcula el factorial de n de forma recursiva."""
    if n <= 1:
        return 1
    return n * factorial(n - 1)
''')

# texto.py
with open('utilidades/texto.py', 'w') as f:
    f.write('''
def capitalizar_palabras(texto):
    """Capitaliza cada palabra del texto."""
    return ' '.join(palabra.capitalize() for palabra in texto.split())

def contar_vocales(texto):
    """Cuenta las vocales en un texto."""
    vocales = 'aeiouáéíóuAEIOUÁÉÍÓÚ'
    return sum(1 for c in texto if c in vocales)
''')

print("Paquete 'utilidades' creado correctamente.")
print(os.listdir('utilidades'))
````

````python
# Importar el paquete propio
import utilidades

# Usar funciones del módulo numeros
print(utilidades.es_par(4))       # True
print(utilidades.es_par(7))       # False
print(utilidades.factorial(6))    # 720

# Usar funciones del módulo texto
print(utilidades.capitalizar_palabras("hola mundo desde python"))
print(utilidades.contar_vocales("programación en python"))
````

````python
# También podés importar directamente desde el submódulo
from utilidades.numeros import factorial
from utilidades.texto import contar_vocales

print(factorial(10))
print(contar_vocales("universidad tecnológica nacional"))
````

## 6. Ver qué contiene un módulo

Podés inspeccionar cualquier módulo con `dir()` y `help()`.

````python
import math

# Ver todo lo disponible en el módulo
publico = [nombre for nombre in dir(math) if not nombre.startswith('_')]
print(publico)
````

````python
# Ver la documentación de una función
help(math.log)
````

````python
# Ver de dónde viene un módulo instalado
import requests
print(requests.__file__)
print(requests.__version__)
````

## Resumen

| Concepto | Descripción |
|----------|-------------|
| **Módulo** | Archivo `.py` con código reutilizable |
| **Paquete** | Carpeta con `__init__.py` que agrupa módulos |
| `import x` | Importa el módulo completo, accedés con `x.algo` |
| `from x import y` | Importa solo `y`, accedés directamente |
| `import x as z` | Importa con alias `z` |
| `pip install` | Instala paquetes de terceros desde PyPI |
| `__init__.py` | Define qué expone el paquete al importarse |

### Paquetes de terceros muy usados

| Paquete | Uso |

| `matplotlib` | Gráficos |
